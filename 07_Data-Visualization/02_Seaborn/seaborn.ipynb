### How to create multiple types of plot with seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.get_dataset_names()
tips=sns.load_dataset("tips")
tips.head()

sns.set_theme()
#relplot
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    size="size",
    style="smoker"
)
```

```python
#lineplot
flights=sns.load_dataset("flights")
flights.head()
```

```python
sns.lineplot(
    data=flights,
    x="year",
    y="passengers",
    #size="passengers"
)
```

<i style="color:red">A shaded area around the line usually represents the confidence interval (CI) around the estimated mean.</i>

### Categorical Plots in Seaborn
- Bar Plots
    - eg., tips dataset.
        - Over the days,how much tips has given by the female and male customer.
- Box Plots
    - eg.,tips dataset.
        - Over the days,what tip value that has been paid.

```python
#Bar plot
tips.head()
```

```python
sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

```python
#box plot
sns.boxplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

### Distribution plots in Seaborn
- Histogram

```python
penguins=sns.load_dataset("penguins")
penguins.head()
```

```python
sns.histplot(
    data=penguins,
    x="body_mass_g",
    bins=35
)
```

### Matrix plot in seaborn
- Heatmaps
    - Plot rectangular data as color-encoded matrix
    - used when data is in 2D array

```python
#Heatmaps
flights.head()
```

```python
flights_pivot=flights.pivot(index="month",columns="year",values="passengers")
```

```python
sns.heatmap(
    flights_pivot,
    cmap="coolwarm",
    annot=True,
    fmt="d"
)
plt.title("Passenger heatmap")
plt.tight_layout()
```

### Seaborn best practices
- Whenever we have to create plot,we are gonna use seaborn,individual plot will be done by using seaborn and if u want to customization or to change the detail at minut level inside it,It can be perform with the help of matplotlib.
- To control the low level details or to add animation elements,matplotlib will handle it.

```python
fig,ax = plt.subplots()

sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    marker="o",
    errorbar=None,
    ax=ax
)
ax.set_title("avg bill value for each day")
ax.set_xlabel("Days")
ax.set_ylabel("Total Bill Amount")
fig.tight_layout()
```
