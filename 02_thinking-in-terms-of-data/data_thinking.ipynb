### Load data

```python
import json
def load_data(filename):
    with open(filename,"r") as f:
        data=json.load(f)
    return data
data=load_data("amazon_store_data.json")
print(data)
```

### Clean the data

```python
def clear_data(data):
    text_to_num={"one":1,"two":2,"three":3,"four":4,"five":5}
    cleaned_data=[]
    unique_user=set()
    for user in data:
        #clean ratings
        raw_rating=str(user["rating"]).strip().lower()
        if raw_rating in text_to_num:
            raw_rating=text_to_num[raw_rating]
        user["rating"]=raw_rating

        #handle missing vals
        raw_age=user.get("age")
        if(raw_age == None):
            user["age"]=None

        #deduplication
        if user["name"] in unique_user:
            continue
        unique_user.add(user["name"])
        cleaned_data.append(user)
    return cleaned_data
```

```python
clear_data(data)
```

### Getting meaningful insights from data
- whats is average rating?
- Whats the % of users gave poor rating(<3)?

```python
def get_insight(data):
    #average rating
    tot_rating=0
    for user in data:
        tot_rating += float(user["rating"])
    print(f"Average rating:{tot_rating/len(data)}")

    #percentage of user with poor rating
    poor_rating=0
    for user in data:
        if float(user["rating"])<3.0:
            poor_rating +=1
    print(f"Percentage of user with poor rating:{poor_rating/len(data)*100}%")
```

```python
get_insight(data)
```

### Build a recommendation feature(Rule based,not ML)
- if user rating>=4,recommend same brand products.
- if user rating<4,recommend different brand products.

```python
def get_recommendation(data):
    recommendation=[]
    
    
    for user in data:
        curr_recomm={}
        curr_recomm["name"]=user["name"]
        if float(user["rating"])>=4:
            curr_recomm["brand"]="Apple"
        else:
            curr_recomm["brand"]="Samsung"
        recommendation.append(curr_recomm)
    return recommendation
```

```python
get_recommendation(data)
```
