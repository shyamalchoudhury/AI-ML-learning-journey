## Handle missing Values

- isnull()/isna()[alias of isnull()]
- isnull().sum()
    - how it is helpful??
- dropna()
- fillna(value)
- ffill()
    - forward fill:It fills missing values (NaN) using the previous valid value in the column or row.
- bfill()
    - backward fill:It fills missing values (NaN) using the next valid value in the column or row.

```python
#Cleaning Data-missing values

df=pd.read_csv("raw_data.csv")

#detect empty val
df.isnull()
#or
df.isna()

#counting the number of missing value per column
df.isnull().sum()

#drop all the null value from our data
df.dropna()#row drop
df.dropna(axis=1)#col drop

#fill the missing value with particular value
df.fillna(0)

#fill the missing value of the column-age and replace them with mean value of the age
age_mean=df["age"].mean()
df["age"].fillna(age_mean)# even the null data is retained,the original data remain unchanged

#To make the change in the original data,copy the original data and store it in the other variable
cleaned_data=df.copy()
age_mean=cleaned_data["age"].mean()
cleaned_data["age"]= cleaned_data["age"].fillna(age_mean)
cleaned_data

#Forward Fill and Backward Fill
df.ffill()
df.bfill()
df
```

## Feature Engineering(Transforming Data)-transforming data to create some neew columns or features that improves the performance of our models.
- apply()
    - used to apply a function to each element, row, or column of a DataFrame or Series.
    - it is a HOF(High Order Function).It means it accepts another function as an argument.
- map()
    - used to transform each value in a Series using a function, dictionary, or another mapping.
- assign()
    - used to create new columns or assign values to columns in a DataFrame. It returns a new DataFrame rather than           modifying the original DataFrame by default.
- replace(old,new)
    - used to replace specific values in a Series or DataFrame with new values.
- rename(columns={old:new})
    - used to change the names of rows (index labels) or columns in a DataFrame.
- sort_values()
    - Sorts rows based on the values in a particular column.
- sort_index()
    - Sorts rows according to their index labels.
- reset_index()
    - Resets the row index to the default sequence 0, 1, 2, ....
- rank()
    - Assigns ranks to values in a column.
    - methods in rank()
        - "average"	Average rank for tied values (default)
        - "min"	Assign the lowest rank to tied values
        - "max"	Assign the highest rank to tied values
        - "first"	Rank ties in their order of appearance
        - "dense"	Like "min", but without gaps in ranks
- Reorder columns
    - you can reorder columns by passing the desired column order inside double square brackets [[ ]].

```python
#feature Engineering

#creating a new column where we can calculate tax on an existing data
#apply()
df2=df.copy()
df2["tax"]=df2["income"].apply(lambda x:"20%" if x >= 60000 else "10%")
df2

#map()
gender_map={"Male":"M","Female":"F","Unknown":"U"}
df2["gender"]=df2["gender"].map(gender_map)
df2

#assign()
df2=df2.assign(new_income=df2["income"]*1.1)

#replace()
df2["country"]=df2["country"].replace("USA","United States")
df2

#rename()
df2.columns=["Id","Name","Age","Coutry","Gender","Income","Tax","New_income"]
df2=df2.rename(columns={"Income":"Salary","New_income":"New_salary"})
df2.rename(index={1:"First"})
df2
#sort_values()
df2.sort_values("Salary")
df2.sort_values("Salary",ascending=False)
sorted_data=df2.sort_values(["Salary","Age"])
sorted_data

#sort_index()
sorted_data.sort_index()

#reset_index()
#it preserved the original index values,therefore a new column is created named as index
sorted_data.reset_index()
#if u dont want to preserved the original data
sorted_data.reset_index(drop=True)


#rank()
sorted_data["Ranking"]=sorted_data["Salary"].rank(ascending=False,method="dense")
sorted_data

#reorder_column
sorted_data[["Id","Name","Age","Coutry","Gender","Tax","Salary","New_salary","Ranking"]]
sorted_data=sorted_data.rename(columns={"Coutry":"Country"})
sorted_data
```

# Grouping and Aggregation of Data
- groupby()
    - used to group rows with the same value and perform operations such as sum(), mean(), count(), and max() on each group.
- mean/sum/max/min/count etc.
- agg()
    - used to apply multiple summary functions to a DataFrame or to groups created using groupby().

```python
df
```

```python
df.groupby("country")["income"].mean()
df.groupby("country")["income"].min()
df.groupby("country")["income"].max()
df.groupby("country")["income"].count()

df.groupby("gender")["income"].mean()
df.groupby("gender")["income"].min()
df.groupby("gender")["income"].max()

#agg() or aggregate()
df.groupby("country")["income"].agg(["mean","min","max"])
df.groupby("country")["income"].aggregate(["mean","min","max"])
#renaming the column
df.groupby("country")["income"].aggregate(avg_salary="mean",minimum="min",maximum="max")

#Applying different aggregate function to different columns
df.groupby("country").agg({
    "income":"max",
    "age":"mean"
})
df.groupby("country").agg(
    max_salary=("income","max"),
    avg_age=("age","mean")
)
```

# Important Operation in Pandas
- melt
    - wide data to long format data
    - melt(id_vars,value_vars,var_name,value_name)
- Pivot
    - long data to wide format data
    - pivot(index,columns,values)

```python
#melt & pivot
import pandas as pd
sales={
    "country":["USA","India","Russia","China"],
    "year":[2021,2020,2023,2024],
    "sales":[100,120,90,130],
    "profit":[20,40,10,43],
    "C_code":[12,91,23,45]
}
df=pd.DataFrame(sales)
melted_df=df.melt(
    id_vars=["country","year"],
    value_vars=["sales","profit","C_code"],
    var_name="matrics",
    value_name="values"
)
original=melted_df.pivot(
    index=["country","year"],
    columns="matrics",
    values="values"
)
```

# Merging and Joining Data

```python
#Merging and Joining data
df_customers=pd.DataFrame(
    {
        "customer_id":[1,2,3,4],
        "name":["Shyamal","Shivam","Soubam","Akash"]
    }
)
df_order=pd.DataFrame({
    "order_id":[101,102,103,104],
    "customer_id":[2,1,4,5],
    "amount":[250,120,340,530]
})
```

```python
#inner join
pd.merge(df_customers,df_order,on="customer_id")
#left join
pd.merge(df_customers,df_order,on="customer_id",how="left")
#right join
pd.merge(df_customers,df_order,on="customer_id",how="right")
#outer join
pd.merge(df_customers,df_order,on="customer_id",how="outer")
```

## Data Concatenation 
- Stacking DataFrames on top of each other(rows) or side by side (columns)
- how it is helpful?
    - Suppose we are fetching data through API calls or retrieving data from a database. Sometimes, due to rate limits or other restrictions, we cannot fetch the entire dataset at once. In such cases, we fetch the data in smaller batches and store each batch in a separate DataFrame.
Eventually, we want to combine all these DataFrames into a single DataFrame. We use data concatenation in Pandas to combine these separate DataFrames into one dataset.

```python
#Concatenation
df1=pd.DataFrame({
    "id":[1,2,3],
    "name":["Shyamal","Shivam","Soubam"],
})
df2=pd.DataFrame({
    "id":[4,5,6],
    "name":["Akash","Arup","Bikash"],
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2])#row
pd.concat([df1,df2],ignore_index=True)#overwriting the original index
```

```python
pd.concat([df1,df2],axis=1) #col
```
