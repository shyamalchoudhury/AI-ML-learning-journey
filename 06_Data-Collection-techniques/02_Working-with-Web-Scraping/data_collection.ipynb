## NOTE:
- In the context of Web scrapping,GET is going to be the most relevant for us.
- We can also use "try-except" blocks to handle exception(s) in our code.This are industry standards

```python
import requests

try:
    response = requests.get("https://example.com", timeout=10)

    response.raise_for_status()

    print("Webpage retrieved successfully!")
    print(response.status_code)

except requests.exceptions.HTTPError as e:
    print("HTTP Error:", e)

except requests.exceptions.RequestException as e:
    print("Request Error:", e)
```

### Headers(Useful in Web scraping)
- Websites often expect certain headers, like “User-Agent” that lets them know that the request is coming from real browser. This can help us avoid getting blocked when sending automated requests.


```python
import requests

url = "https://example.com"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

response.raise_for_status()

print(response.status_code)
print(response.text)
```

### time.sleep()
-  We can also introduce delays using time.sleep() in scraping to avoid hitting servers too fast.

```python
import requests
import time

urls = [
    "https://www.scrapethissite.com/pages/ajax-javascript/#2015",
    "https://www.scrapethissite.com/pages/ajax-javascript/#2014",
    "https://www.scrapethissite.com/pages/ajax-javascript/#2013",
    "https://www.scrapethissite.com/pages/ajax-javascript/#2012",
    "https://www.scrapethissite.com/pages/ajax-javascript/#2011",
    "https://www.scrapethissite.com/pages/ajax-javascript/#2010"
]

for url in urls:
    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()

        print(f"Successfully scraped: {url}")

        time.sleep(2)  # Wait for 2 seconds

    except requests.exceptions.RequestException as e:
        print(f"Error: {e}")
```

```python
from bs4 import BeautifulSoup
#reading the scraped_data
with open("scraped_data/data1.html","r") as f:
    html_content=f.read()
#now we are going to create soup.so beautiful soup is a class,and we are gonna create object which we call soup
soup=BeautifulSoup(html_content,"lxml")
soup
```

### data is parsed,but how to start extracting useful information.
- For example,lets dive into the the site that we scraped.
- lets create a table where we are having countries name and its population data.
- but there are too many unnecessary data in the sites,inspect it and look for the tag where the countries have mentioned.
- Therefore,you may noticed that the countries name are stored in <.h3>...<./h3> tag,with common class as "country-name".
- Parse Tree:
  
        countries
         /     \
      name     population
        \           \
      h3tags       div class="country-info"
      /                        \
      class="country-name"       span class="country-population"
- Methods in Beautiful Soup to extract information.
    - .decompose()=Remove unwanted HTML elements.
    - .find()=find first matching element.
    - .find_all()=Finds all matching elements
    - .find_parent()=Finds a matching ancestor
    - .find_parents()=Finds all matching ancestor
    - .find_next()=Finds the next matching element in document order
    - .find_previous()=Finds the previous matching element
    - access thru CSS selectors
        - .select()=Finds all elements matching a CSS selector
        - .select_one()=Finds the first element matching a CSS selector
    - access thru attributes
        - get()=Extract an attribute
        - .attrs=Returns all attributes as a dictionary
    - Navigating DOM
        - .parent
        - .children
        - .descendants
        - .next_sibling()
        - .previous_sibling()

```python
import pandas as pd
all_h3=soup.find_all("h3")
data=[]
for h3 in all_h3:
    Country=h3.get_text(strip=True) #to access the content of the h3 tag:get_text()
    #print(h3.find_parent("div")["class"])
    population=h3.find_next("div").select_one("span.country-population").get_text(strip=True)
    data.append({
        "Country":Country,
        "Population":population
    })
df=pd.DataFrame(data)
display(df)
```

```python
df.to_csv("cleaned_data/data.csv",index=False)
```
