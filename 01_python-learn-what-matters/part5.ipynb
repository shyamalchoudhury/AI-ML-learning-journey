# File I/O 

## write
- Creates new file or overwrites existing file.

```python
with open("data.txt","w")as f:
    content="This is my first File"
    num_of_character=f.write(content)
    print(content)
    print(f"File contain {num_of_character} number of character(including space)")

print("\n")

with open("second_file.txt","w") as f:
    content="This text file contain JSON data"
    num_of_char=f.write(content)
    print(content)
    print(f"File contain {num_of_char} number of character(including space)")
```

## Deleting a file

```python
import os
os.remove("sencond_file")
os.remove("second_file")
```

## Append
- Add content at the end of the fie

```python
with open("third_file.txt","a") as f:
    content="\nThis line has been written using append command"
    f.write(content)
with open("third_file.txt","r") as f:
    c=f.read()
    print(c)
```

## read
- Opens file for reading(default).Error if file doesn't exist

```python
with open("third_file.txt","r") as f:
    content=f.read()
    print(content)
    print(f"Length of content:{len(content)}")
```

## Copying the content of a file to another file

```python
with open("third_file.txt","r") as f1,open("copied_file.txt","w") as f2:
    content=f1.read()
    f2.write(content)
with open("copied_file.txt","r") as f2:
    content=f2.read()
    print(content)
```

## Exception Handling
- Three Block
  - try,except,else,finally

```python
x=10
y=0
print(x/y)
```

```python
x=10
y=0
try :
    print(x/y)
except ZeroDivisionError as e:
    msg=str(e)
    print(msg)
else:
    print("denominaotor is not 0")
finally:
    print(1/2)
```

# List Comprehension

## Basic syntax

```python
num=[x for x in range(1,6)]
print(num)
```

## with if-condition

```python
evens=[x for x in range(1,11) if x%2==0]
print(evens)
```

## with if-else condition

```python
nums=[x for x in range(1,12)]
print(nums)
labels=[(x,"Even" if x%2==0 else "Odd") for x in nums]
print(labels)
```

## representing in the form of dictionary

```python
labels={x:"Even" if x%2==0 else "Odd" for x in nums}
print(labels)
```

# Working with JSON module

## Converting python objects to json string

```python
import json
data={
    "name":"Shyamal",
    "Age":24,
    "Course":"MCA",
    "Subject":["IOT","NSAC","MOS"]
}
json_string=json.dumps(data)
print(data)
print(data["name"])
print(json_string)
```

## Converting json string to python objects

### To access the json string,convert back to python dictionary

```python
python_objects=json.loads(json_string)
print(python_objects)
print(python_objects["name"])
```

# Writing JSON from a file

```python
import json
data={"name":"Shyamal","City":"Guwahati"}
with open("data.json","w") as f:
    json.dump(data,f)
```

## Reading json from a file

```python
import json
with open("data.json","r") as f:
    data=json.load(f)
print(data)
```

### create a program that:
- Opens a file "names.txt" in write mode
- Writes 5 names (one per line)entered by the user
- THEN OPENS THE SAME FILE IN READ MODE AND PRINT ALL NAMES

```python
with open("names.txt","w") as f:
    for i in range(5):
        name=input(f"Enter name{i+1}:")
        f.write(name +"\n")
print("\n")
with open("names.txt","r") as f:
    print("Names in the file:")
    for name in f:
        print(name.strip())
```
