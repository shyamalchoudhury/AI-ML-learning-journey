# Create a function to compute factorial of a number

```python
def factorial(n):
    fact=1
    i=1
    while(i<=n):
        fact *=i
        i +=1
    return fact
factorial(3)
```

# Create a function to return lagest of 3 numbers - a,b & c

```python
def largest(a,b,c):
    if a>b and a>c:
        return a
    elif b>c:
        return b
    else:
        return c
largest(5,6,9)
```

```python
def print_digit(n):
    count=0
    while(n!=0):
        digit=n%10
        count +=1
        print(f"The digit are:{digit}")
        n = n // 10
    print(f"Number of digit in the number is:{count}")
n=int(input("Enter the number:"))
print_digit(n)
```

```python
def digit_sum(n):
    original=n
    sum =0
    while(n!=0):
        digit=n%10
        sum +=digit
        n=n //10
    print(f"The sum of each digit of {original} is {sum}")
n=int(input("Enter the number"))
digit_sum(n)
```

```python
def calculator(a,b,operation):
    if operation=="+":
        return a+b
    elif operation=="-":
        return a-b
    elif operation=="*":
        return a*b
    elif operation=="/":
        return a/b
    else:
       print("Invalid Operation")
a=int(input("Enter the first number:"))
b=int(input("Enter the Second number:"))
operation=input("what operation u want to perform:")
print(calculator(a,b,operation))
```
