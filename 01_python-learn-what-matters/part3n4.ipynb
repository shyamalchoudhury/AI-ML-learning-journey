## Instance Methods 

```python
class Student:
    school_name="ABC school"
    @classmethod
    def standard(cls,new_name):
        cls.school_name=new_name
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def display(self):
        print(f"Name:{self.name},Age:{self.age}")
s1=Student("Shyamal",18)
s2=Student("Shyam",13)
print(Student.school_name)

Student.standard("Merry")
print(Student.school_name)

s1.display()
s2.display()
```

```python
class Internship:
    #college_name="Royal Global University"
    role_name="SDE 1"
    
    @staticmethod
    def company():
        return "Accenture"
        
    @classmethod
    def role(cls,new_role_name):
        cls.role_name=new_role_name

        
    def __init__(self,name,age,month):
        self.name=name
        self.age=age
        self.month=month
        
    def display(self):
        print(f"\tName:{self.name} \n \tAge:{self.age}\n \tMonth:{self.month}")
        
class Employee(Internship):
    def __init__(self,name,duration):
        self.name=name
        self.duration=duration

        
class Bank:
    def __init__(self,salary):
        self.__salary=salary
    def get_salary(self):
        return self.__salary
    def set_salary(self,new_salary):
        self.__salary=new_salary

b=Bank(100000)


intern1=Internship("Shyam",21,6)
print(Internship.company())
print(f"Role:{Internship.role_name}")
intern1.display()
print(f"\tSalary:{b.get_salary()}")

b.set_salary(120000)

intern2=Internship("Shyamal",24,6)
Internship.role("SDE 2")
print(f"Role:{Internship.role_name}")
intern2.display()
print(f"\tSalary:{b.get_salary()}")

emp1=Employee("Sena","2years")
emp2=Employee("Senapati","4years")
emp3=Employee("Sen","3years")

print(f"\n{emp1.company()}")
print(f"Role:{emp1.role_name}")
print(f"\tName:{emp1.name}\n \tDuration:{emp1.duration}")
print(f"\n \tName:{emp2.name}\n \tDuration:{emp2.duration}")
print(f"\n \tName:{emp3.name}\n \tDuration:{emp3.duration}")
```
